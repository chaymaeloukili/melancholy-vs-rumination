## 2. Main result of mood after the experience per sample 

```python
pd.crosstab(df['MOODCHANGE'], df['sample'])
```

```python
(pd.crosstab(df['MOODCHANGE'], df['sample'], normalize="columns") * 100).round(1)
```

**What I found**
- S1 (Finland, convenience) is not representative: people joined via a media call, so many of them like sad music.
- S2 (Finland, quota) was selected to match the UK sample in gender and in how often people listen to sad music.
- S1 and S2 look similar: most people felt better after the experience (S1 79.2%, S2 73.5%), and only few felt worse.
- S3 (UK) is very different: only 37.8% felt better, almost half (47.8%) felt the same, and 14.4% felt worse.
- So in the UK sample, sad music did not make most people feel worse – but fewer people felt better, and more stayed neutral.
- Because S2 was built to be similar to the UK sample, the difference is probably not only about "fans" of sad music.

**Open questions (ideas, not tested)**
- Culture: do people experience or talk about sad music differently?
- Language of the survey: the question was translated, so "better" may not mean exactly the same in Finnish and English.
- Language of the songs: if a song is not in your mother tongue, maybe you focus more on the melody than on the lyrics. This could change how strongly the music affects you.
- Chance: still needs to be checked with statistics.

Note: these are self-reports (how people remember and feel it), not measurements of the brain.

**What I found**
- In S3 (UK) and S2 (Finland, quota) the same pattern appears: the more often people listen to sad music, the more often they felt better and the less often they felt worse.
- People who don't listen to sad music voluntarily mostly did not feel better (S3: 9.4%, S2: 17.6%), and often felt worse (S3: 28.1%, S2: 52.9%). These groups are very small (n = 32 and n = 17).

**Open questions (not testable with this data)**
- Direction: does listening often make people feel better, or do people who feel better afterwards simply listen more often (and others avoid it)?
- Outlet vs. habituation: maybe sad music works as an outlet for people who listen often. Pure habituation would rather predict more "same".

**S1 (Finland, fans)**

**Summary step 4**

- Being a "fan" adds little beyond listening frequency.
- The country difference remains: at the same listening frequency, Finland ≈ 80% better, UK ≈ 40%.
- This is a link, not proof of cause. Small groups ("I don't listen": n = 12–32) are uncertain.

## 6. General reasons for listening to sad music: Worse vs. Better

```python
# R1-R24: general reasons for listening to sad music (checkbox: 1 = ticked)
reason_cols = [c for c in df.columns if c.startswith("R") and c[1:].isdigit()]
reason_labels = pd.read_csv(RAW_DIR / "REASON_label.csv", header=None)[0].tolist()

print(len(reason_cols), len(reason_labels))   # expected: 24 24
```

```python
# Number of people who ticked each reason, per sample and mood group
reason_counts = df.groupby(['sample', 'MOODCHANGE'], observed=True)[reason_cols].sum().T
reason_counts.insert(0, "reason", reason_labels)
reason_counts
```

```python
# Share of people who ticked each reason, in percent
reason_pct = (df.groupby(['sample', 'MOODCHANGE'], observed=True)[reason_cols].mean() * 100).round(1).T
reason_pct.insert(0, "reason", reason_labels)
reason_pct
```

```python
# How many people are in each group (n)
pd.crosstab(df['MOODCHANGE'], df['sample'])
```

```python
%pip install jinja2
# Difference in percentage points: Better minus Worse
better = reason_pct.drop(columns="reason").xs('Better', axis=1, level='MOODCHANGE')
worse = reason_pct.drop(columns="reason").xs('Worse', axis=1, level='MOODCHANGE')

reason_diff = (better - worse).round(1)
reason_diff.insert(0, "reason", reason_labels)
reason_diff = reason_diff.sort_values("S1_Finland_convenience")

# Colours: blue = more often in Better, red = more often in Worse, white = no difference
sample_cols = ["S1_Finland_convenience", "S2_Finland_quota", "S3_UK_representative"]
reason_diff.style.background_gradient(cmap="RdBu", vmin=-50, vmax=50, subset=sample_cols).format(precision=1, subset=sample_cols)
```

This table shows how often each reason was ticked in each sample overall (not Better vs. Worse), to compare the samples with each other.

```python
# Share of people per sample who ticked each general reason (in percent), without splitting into Better/Worse
reason_by_sample = (df.groupby('sample')[reason_cols].mean() * 100).round(1).T

# Add the reason texts as the first column, so you can read what each R number means
reason_by_sample.insert(0, "reason", reason_labels)

# Show the table
reason_by_sample
```

**What I found (step 6: general reasons)**

*Better vs. Worse (within each sample)*
- In all three samples, people who felt better had ticked almost every reason more often than people who felt worse. They seem to have a closer, more conscious relationship with sad music.
- Robust in all samples: to relax and calm down, because it is beautiful, to improve my mood, to comfort me, to get in touch with my emotions, to channel my emotions.
- "To avoid negative thoughts and feelings" hardly differs (close to 0) and is rarely ticked overall (5–9%). Using sad music to *allow* feelings seems more linked to feeling better than using it to *push feelings away*.
- No reason is clearly typical for the Worse group. The Worse group simply has fewer reasons. Only "Other" is more common in the UK Worse group (−12), so their reasons remain unknown.
- Some large differences only appear in S2 (e.g. "congruent with my mood" +35). S2 has only 30 Worse cases, so these are hints, not findings.

*Samples compared (overall)*
- My first idea was that S2 listens more consciously. The overall table does not support this: for almost every reason S1 > S2 > S3.
- UK participants ticked far fewer reasons overall (e.g. "channel my emotions": S1 62%, S2 51%, S3 24%). Sad music seems to play a smaller role in their lives – which fits that many of them felt "the same".

*Note:* R1–R24 are general reasons, not reasons for the specific experience. The results show links, not causes.

## 7. Attitudes towards sad music: Worse vs. Better
These are general attitudes on a scale from 1 (not at all) to 7 (very much), so I compare average values instead of percentages.

```python
# ASM1-ASM25: attitudes towards sad music (scale 1-7)
asm_cols = [c for c in df.columns if c.startswith("ASM")]
asm_labels = pd.read_csv(RAW_DIR / "ASM_label.csv", header=None)[0].tolist()

# The labels start with a code like "F2-10 ", so I remove everything before the first space
asm_labels = [label.split(" ", 1)[1] for label in asm_labels]

print(len(asm_cols), len(asm_labels))   # expected: 25 25
```

```python
# Average agreement (1-7) per sample and mood group
asm_mean = df.groupby(['sample', 'MOODCHANGE'], observed=True)[asm_cols].mean().round(2).T
asm_mean.insert(0, "statement", asm_labels)
asm_mean
```

```python
# Difference in scale points: Better minus Worse (blue = higher in Better, red = higher in Worse)
better = asm_mean.drop(columns="statement").xs('Better', axis=1, level='MOODCHANGE')
worse = asm_mean.drop(columns="statement").xs('Worse', axis=1, level='MOODCHANGE')

asm_diff = (better - worse).round(2)
asm_diff.insert(0, "statement", asm_labels)
asm_diff = asm_diff.sort_values("S1_Finland_convenience")

asm_diff.style.background_gradient(cmap="RdBu", vmin=-2, vmax=2, subset=sample_cols).format(precision=2, subset=sample_cols)
```

```python
# Average agreement (1-7) per sample, without splitting into Better/Worse
asm_by_sample = df.groupby('sample')[asm_cols].mean().round(2).T
asm_by_sample.insert(0, "statement", asm_labels)
asm_by_sample
```

**What I found (step 7: attitudes)**
- Robust in all samples: people who felt better agree more that sad music uplifts them, relaxes them and makes them grateful, and they sometimes seek sadness on purpose. People who felt worse rather avoid sad music when they are sad, and it reminds them of their own tragedies (fits rumination).
- Only in Finland: people who felt worse agree much more that sad music intensifies negative feelings, makes them anxious or irritates them.
- In the UK, the link for "intensifies my negative feelings" is reversed (+0.64): people who felt better agree more. Intensifying feelings might be experienced as an outlet there – the same statement seems to mean something different in the two countries.
- "I regulate my negative feelings by listening to sad music" (ASM3) is not consistent across samples (+0.26 / +1.32 / −0.50).
- UK: people who empathize easily with sad music felt worse more often (−1.28) – maybe they take on the sadness more strongly.
- Values are differences in scale points (1–7). S2 has only 30 Worse cases, so its large values are uncertain.

**Attitudes compared between samples (overall)**
- UK agrees much less with positive statements (e.g. "gives me strength": 3.16 vs. 5.36 in S1) and much more with negative ones (e.g. "makes me anxious": 4.71 vs. 2.39).
- So in the UK sad music is generally seen more critically – which fits that fewer UK participants felt better after the experience.
- S1 and S2 are again very similar.
- "Intensifies my negative feelings" is similar everywhere (~4): all samples agree equally that sad music intensifies feelings, but they seem to judge this differently.
- Limitation: differences between countries could also come from translation or from how people use rating scales.
- The attitudes mirror the general reasons from step 6 (e.g. "to relax" ↔ "relaxes me", "to comfort me" ↔ "gives me strength"). Two different types of questions (checkboxes and 1–7 scales) show the same picture, which makes the results more convincing.
