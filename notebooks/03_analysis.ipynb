## 2. Main result of mood after the experience per sample 

```python
pd.crosstab(df['MOODCHANGE'], df['sample'])
```

```python
(pd.crosstab(df['MOODCHANGE'], df['sample'], normalize="columns") * 100).round(1)
```

**What I found**
- S1 (Finland, convenience) is not representative: people joined via a media call, so many of them like sad music.
- S2 (Finland, quota) was selected to match the UK sample in gender and in how often people listen to sad music.
- S1 and S2 look similar: most people felt better after the experience (S1 79.2%, S2 73.5%), and only few felt worse.
- S3 (UK) is very different: only 37.8% felt better, almost half (47.8%) felt the same, and 14.4% felt worse.
- So in the UK sample, sad music did not make most people feel worse – but fewer people felt better, and more stayed neutral.
- Because S2 was built to be similar to the UK sample, the difference is probably not only about "fans" of sad music.

**Open questions (ideas, not tested)**
- Culture: do people experience or talk about sad music differently?
- Language of the survey: the question was translated, so "better" may not mean exactly the same in Finnish and English.
- Language of the songs: if a song is not in your mother tongue, maybe you focus more on the melody than on the lyrics. This could change how strongly the music affects you.
- Chance: still needs to be checked with statistics.

Note: these are self-reports (how people remember and feel it), not measurements of the brain.

## 7. Attitudes towards sad music: Worse vs. Better
These are general attitudes on a scale from 1 (not at all) to 7 (very much), so I compare average values instead of percentages.

```python
# ASM1-ASM25: attitudes towards sad music (scale 1-7)
asm_cols = [c for c in df.columns if c.startswith("ASM")]
asm_labels = pd.read_csv(RAW_DIR / "ASM_label.csv", header=None)[0].tolist()

# The labels start with a code like "F2-10 ", so I remove everything before the first space
asm_labels = [label.split(" ", 1)[1] for label in asm_labels]

print(len(asm_cols), len(asm_labels))   # expected: 25 25
```

```python
# Average agreement (1-7) per sample and mood group
asm_mean = df.groupby(['sample', 'MOODCHANGE'], observed=True)[asm_cols].mean().round(2).T
asm_mean.insert(0, "statement", asm_labels)
asm_mean
```

```python
# Difference in scale points: Better minus Worse (blue = higher in Better, red = higher in Worse)
better = asm_mean.drop(columns="statement").xs('Better', axis=1, level='MOODCHANGE')
worse = asm_mean.drop(columns="statement").xs('Worse', axis=1, level='MOODCHANGE')

asm_diff = (better - worse).round(2)
asm_diff.insert(0, "statement", asm_labels)
asm_diff = asm_diff.sort_values("S1_Finland_convenience")

asm_diff.style.background_gradient(cmap="RdBu", vmin=-2, vmax=2, subset=sample_cols).format(precision=2, subset=sample_cols)
```

```python
# Average agreement (1-7) per sample, without splitting into Better/Worse
asm_by_sample = df.groupby('sample')[asm_cols].mean().round(2).T
asm_by_sample.insert(0, "statement", asm_labels)
asm_by_sample
```

**What I found (step 7: attitudes)**
- Robust in all samples: people who felt better agree more that sad music uplifts them, relaxes them and makes them grateful, and they sometimes seek sadness on purpose. People who felt worse rather avoid sad music when they are sad, and it reminds them of their own tragedies (fits rumination).
- Only in Finland: people who felt worse agree much more that sad music intensifies negative feelings, makes them anxious or irritates them.
- ~~In the UK, the link for "intensifies my negative feelings" is reversed (+0.64): people who felt better agree more. Intensifying feelings might be experienced as an outlet there – the same statement seems to mean something different in the two countries.~~
- "I regulate my negative feelings by listening to sad music" (ASM3) is not consistent across samples (+0.26 / +1.32 / −0.50).
- UK: people who empathize easily with sad music felt worse more often (−1.28) – maybe they take on the sadness more strongly.
- Values are differences in scale points (1–7). S2 has only 30 Worse cases, so its large values are uncertain.

**Attitudes compared between samples (overall)**
- UK agrees much less with positive statements (e.g. "gives me strength": 3.16 vs. 5.36 in S1) and much more with negative ones (e.g. "makes me anxious": 4.71 vs. 2.39).
- So in the UK sad music is generally seen more critically – which fits that fewer UK participants felt better after the experience.

- The attitudes mirror the general reasons from step 6 (e.g. "to relax" ↔ "relaxes me", "to comfort me" ↔ "gives me strength"). Two different types of questions (checkboxes and 1–7 scales) show the same picture, which makes the results more convincing.

### 7b. Data quality check: are the UK attitude answers plausible?

```python
# Correlation of "relaxes me" (ASM2) with "irritates me" (ASM10) and "unpleasant feelings" (ASM11), per sample
for s in sample_cols:
    corr = df[df['sample'] == s][asm_cols].corr()
    print(s, corr.loc['ASM2', ['ASM10', 'ASM11']].round(2).tolist())
```

```python
import numpy as np

# All correlations between the 25 ASM items, per sample (each pair counted once)
mask = np.triu(np.ones((25, 25)), k=1).astype(bool)
pairs = {s: df[df['sample'] == s][asm_cols].corr().values[mask] for s in sample_cols}

s1, s2, s3 = sample_cols
print("S1 vs S2:", round(np.corrcoef(pairs[s1], pairs[s2])[0, 1], 2))
print("S1 vs S3:", round(np.corrcoef(pairs[s1], pairs[s3])[0, 1], 2))
print("Share of negative correlations:", {s: round((pairs[s] < 0).mean(), 2) for s in sample_cols})
```

```python
# How different are the average answers of the 25 statements within each sample?
df.groupby('sample')[asm_cols].mean().std(axis=1).round(2)
```

**Result of the check**
- In S1 and S2, "relaxes me" correlates negatively with "irritates me" and "unpleasant feelings" (about −0.4). In S3 (UK) it correlates positively (+0.35 / +0.47), which is not plausible.
- The pattern of all item correlations is almost identical in S1 and S2 (r = 0.96), but S3 does not match it (r = 0.04).
- In S3 the average answers are squeezed towards the middle (SD of item means 0.49 vs. 0.92–1.11 in Finland).
- Possible causes: the items were shown in random order and the UK export may have stored the display order instead of the item number, or acquiescent responding. Not proven.
- → **The UK results for the ASM attitudes are not reliable and are not used as findings.** The Finnish results are not affected.

```python
# Plausibility check for the reasons: do related reasons go together in every sample?
for s in sample_cols:
    corr = df[df['sample'] == s][reason_cols].corr()
    print(s, corr.loc['R8', ['R9', 'R13']].round(2).tolist())
```

**Cross-check: reasons and feelings**

### 8a. Situation: how important was the experience?

```python
def mood_by(column):
    """Show mood change by a situation column: row percentages and group sizes (n)."""
    pct = (pd.crosstab([df['sample'], df[column]], df['MOODCHANGE'], normalize="index") * 100).round(1)
    n = pd.crosstab([df['sample'], df[column]], df['MOODCHANGE'], margins=True)['All']
    pct['n'] = n  # add the group size as an extra column
    return pct
```

```python
mood_by('IMPORTANCE')
```

**What I see (8a: importance)**
- Finland S1: almost no difference, "Better" is 76–83 % in every group.
- Finland S2: only "Not at all" stands out (fewer "Better" 57 %, more "Worse" 20 %). The other groups are all 75–79 %.
- UK: the more important the experience, the more often people felt better (27 % → 59 %). "Worse" does not go down; mainly "Same" gets less.
- UK "Extremely Important" is small (n = 46).
- The direction is unclear: an experience that helped may be remembered as more important.

## 9. Statistical tests

Chi-square test: could the differences in the tables be due to chance?
p < 0.05 means the difference is probably not due to chance. A small p-value does not mean the difference is large.

```python
%pip install scipy
```

```python
from scipy.stats import chi2_contingency

# Test 1: does mood change differ between the samples? (table from step 2)
table = pd.crosstab(df['MOODCHANGE'], df['sample'])
chi2, p, dof, expected = chi2_contingency(table)
print(f"chi2 = {chi2:.1f}, df = {dof}, p = {p:.3g}")
```

**Test 1: mood change between samples**
- chi2 = 243.1, df = 4, p = 2e-51 → not chance.
- Makes sense: the gap is big (Finland 75–80 % "Better", UK 38 %) and there are almost 1,900 answers.
- The test only shows *that* the samples differ, not *why*. Possible reasons: fans in S1, more "not self-chosen" in UK (8a), culture. I can't separate them here.

```python
def chi_test(column):
    """Chi-square test of mood change vs. a column, separately for each sample."""
    for s in sample_cols:
        sub = df[df['sample'] == s]
        table = pd.crosstab(sub[column], sub['MOODCHANGE'])
        chi2, p, dof, expected = chi2_contingency(table)
        n = table.values.sum()
        v = (chi2 / (n * (min(table.shape) - 1))) ** 0.5  # Cramér's V: effect size from 0 to 1
        print(f"{column:10} | {s:24} | p = {p:.3g} | V = {v:.2f}")

for col in ['CHOICE', 'FAM', 'IMPORTANCE', 'DURATION', 'AGO']:
    chi_test(col)
```

**Test 2: situation and mood change (per sample)**
- CHOICE and DURATION: not chance in all three samples, weak to medium effect (V = 0.21–0.35). These are the most robust situation results.
- FAM and IMPORTANCE: not chance in S2 and UK, but not in S1. Weak effect (V around 0.16).
- AGO: no effect in any sample.
- This matches what I saw in the tables in step 8a.
- Careful: I ran 15 tests, so a p-value like 0.02 could still be chance. Some groups are small, which makes the test less exact.

```python
def chi_items(cols, labels):
    """Cramér's V for each 0/1 item vs. mood change, per sample. Empty = not significant (p >= 0.001)."""
    result = pd.DataFrame({"item": labels}, index=cols)
    for s in sample_cols:
        sub = df[df['sample'] == s]
        values = []
        for col in cols:
            table = pd.crosstab(sub[col], sub['MOODCHANGE'])
            if min(table.shape) < 2:      # item never ticked in this sample -> no test possible
                values.append(None)
                continue
            chi2, p, dof, expected = chi2_contingency(table)
            v = (chi2 / (table.values.sum() * (min(table.shape) - 1))) ** 0.5
            values.append(round(v, 2) if p < 0.001 else None)
        result[s] = values
    return result

chi_items(phys_cols, phys_labels)
```

**Test 3: physical reactions (p < 0.001, stricter because of many tests)**
- "Cried a little": no difference in any sample.
- "No physical reaction" and "cried a lot": real but weak differences in S1 and UK (V = 0.12–0.22). Fits 8b: "Same" has the fewest reactions.
- Chills: only in S1, very weak (V = 0.12).
- S2 is smaller, so it is harder to reach p < 0.001. Empty does not mean "no effect", only "not sure".

```python
chi_items(mec_cols, mec_labels)
```

**Test 4: mechanisms (p < 0.001)**
- Only two mechanisms are linked to mood change, and in all three samples: beauty (V = 0.26–0.35) and emotion expressed by the music (V = 0.18–0.26).
- All other mechanisms: no clear difference.
- Memories: not significant. So the hint from 8b ("memories more common in Worse") is not confirmed by the test.

**Test 5: feelings and reasons (p < 0.001)**
- Feelings linked to mood change in all three samples: comfort, relief, peacefulness, sad but elated (V = 0.20–0.36). Step 5 shows the direction: comfort, relief and peacefulness go with "Better".
- Only in Finland: pleasant melancholia, transcendence, anxiety, grief-stricken, depression.
- Reasons: only "because it is beautiful" is significant in all three samples (V = 0.19–0.36).
- Beauty shows up three times: as a mechanism (8b), as a reason (step 6) and in pleasure (8c). This is the most robust theme of my analysis.
- Memories are not significant, neither as a reason (R10) nor as a mechanism (MEC3).
- The test only shows *that* there is a difference, not the direction. The direction comes from steps 5 and 6.

## 10. Conclusion

Looking at who feels better after listening to music they remember as sad, there is a clear difference between Finland and the UK. In the UK "Same" is the most common answer, which means that for most people there the music did not change their mood, neither in a good nor in a bad way. Age does not explain this difference. It is important to keep in mind that sample 1 probably includes many fans of sad music, but its results are still close to the Finnish quota sample.

How often people listen to sad music shows interesting results. The more often people listen to it, the more often they feel better afterwards, and this is true in all three samples. From "Sometimes" on, S1 and S2 are almost the same. So the higher numbers in S1 seem to come mostly from having more frequent listeners among the fans, and not from being a fan as such.

Across all three samples there is a common thread. People whose mood got better felt comfort, relief and peacefulness, and "sad but elated" also stands out in all three samples. For the people who felt worse, the feelings are less clear and not the same in every sample.

The reasons for listening can be very different. "Because it is beautiful" is the only reason that stands out clearly in all three samples. "To avoid negative thoughts" shows no difference at all. People who felt better ticked almost all reasons more often, and in the UK people ticked fewer reasons in general, they seem more reserved in their answers.

The attitude answers (ASM) from the UK are not reliable, so I leave them out of my conclusions. How long ago the experience happened had no effect overall, so I do not take it into account either.

When it comes to physical reactions and mechanisms, only beauty and emotion expressed by the music stand out in all samples. This fits with the reasons, and it may be part of why people listen to this kind of music at all. Beauty shows up again and again in my analysis, as a mechanism, as a reason and in how pleasant the experience felt.

Intensity is lowest in the "Same" group, while "Better" and "Worse" are almost equally high. This means that both groups had a strong experience, and the difference between them is not how strong it was but how it felt. Pleasure is much higher in the "Better" group than in the "Worse" group. This may fit with what I found about the situation, since people who chose the music themselves and listened longer felt better more often. But I cannot say what came first.

Memories did not make a difference, neither as a reason nor as a mechanism.

In general, I cannot draw firm conclusions from this, because the answers are only about emotions, which can also be part of a phase people are going through. People were remembering an experience, and memories can be blurry and inexact. There is no medical data or diagnosis behind the answers, only what people said about themselves. So the results may point in a direction, but they do not prove it. I only found connections, not what causes what or why. For that I would need more data and more information, which probably never really ends, but that is what makes it interesting.

It is also important that the three samples have different kinds of participants. Sample 1 is mostly fans, sample 2 is a quota sample and the UK sample is representative, so the UK can best be compared with sample 2. Some groups were also very small, for example only 22 people in sample 2 who did not know the music. And the attitude answers (ASM) from the UK were not usable, while the other questions in the UK looked fine.

What surprised me most was how different the UK values are from the others, and that age does not play a role. It also stayed with me that people who listen to sad music more often feel better afterwards. Maybe for them it is a lasting way to regulate their feelings, and maybe also a way to allow themselves to cry. What I was actually looking for, and what I find beautiful, is that beauty runs through everything. Even some of the people who felt worse said that beauty was part of their experience. Maybe everyone needs that sometimes, something beautiful to hold on to, even when it is sad.
